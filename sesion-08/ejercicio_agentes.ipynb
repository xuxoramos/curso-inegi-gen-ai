Añadí y ejecuté una celda bootstrap reproducible con 2,000 réplicas:

Efecto estimado: -4.46 puntos de score

El efecto corresponde a la diferencia de medias entre programa = 1 y programa = 0. En esta estimación no se aplicó el factor de expansión.

````python
import numpy as np
from decimal import Decimal, getcontext


getcontext().prec = 28

def diferencia_score(datos):
    score_programa = datos.loc[datos["programa"] == 1, "score"].mean()
    score_control = datos.loc[datos["programa"] == 0, "score"].mean()
    return float(Decimal(str(score_programa)) - Decimal(str(score_control)))

semilla = 2026
replicas = 2000
rng = np.random.default_rng(semilla)
efecto = diferencia_score(encuesta)
efectos_bootstrap = np.empty(replicas)

for replica in range(replicas):
    indices = rng.integers(0, len(encuesta), size=len(encuesta))
    efectos_bootstrap[replica] = diferencia_score(encuesta.iloc[indices])

error_estandar = efectos_bootstrap.std(ddof=1)
intervalo_95 = np.percentile(efectos_bootstrap, [2.5, 97.5])

print(f"Efecto estimado (programa - control): {efecto:.2f} puntos de score")
print(f"Error estándar bootstrap: {error_estandar:.2f}")
print(f"IC bootstrap del 95%: [{intervalo_95[0]:.2f}, {intervalo_95[1]:.2f}]")
````
